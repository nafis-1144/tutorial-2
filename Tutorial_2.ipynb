```python
import numpy as np

array_2x2 = np.array([[1, 2],
                      [3, 4]])

print(array_2x2)
```

```python
random_array = np.random.rand(3, 4)
print(random_array)
```

```python
ones_array = np.ones((2, 3))
zeros_array = np.zeros((2, 3))

print("Array of ones:")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

```python
array_6x3 = np.random.rand(6, 3)
transposed_array = array_6x3.T

print("Original 6×3 array:")
print(array_6x3)

print("\nTransposed 3×6 array:")
print(transposed_array)
```

```python
array_a = np.array([1, 2, 3])
array_b = np.array([4, 5, 6])

dot_product = np.dot(array_a, array_b)

print("Dot product:", dot_product)
```
